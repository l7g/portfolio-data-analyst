# Data cleaning and merge

## 0. Setup

```python
import glob
from pathlib import Path

import numpy as np
import pandas as pd

data_dir = Path.cwd().parent / "data" / "raw" / "air_quality"
files_list = list(glob.glob(str(data_dir / "*.csv")))

station_info = {
    "CENCA1": {
        "comune": "Cagliari",
        "via": "Via Cadello",
        "tipo": "traffico"
    },
    "CENM01": {
        "comune": "Monserrato",
        "via": "Via Sant'Angelo",
        "tipo": "fondo"
    },
    "CENQU1": {
        "comune": "Quartu Sant'Elena",
        "via": "Via Perdalonga",
        "tipo": "fondo"
    }
}

print(f"Found {len(files_list)} files in {data_dir}")
```

## 1. Single file inspection

```python
random_numbers = np.random.randint(0, len(files_list), size=2)

pilot_file = files_list[random_numbers[0]]
compare_file = files_list[random_numbers[1]]
print(pilot_file)
print(compare_file)
```

```python
pilot_df = pd.read_csv(pilot_file, sep=";", decimal=",")
compare_df = pd.read_csv(compare_file, sep=";", decimal=",")
names = ["pilot_df", "compare_df"]

for df in zip([pilot_df, compare_df], names):
    print(f"--- {df[1]} ---")
    print(f"\nShape: {df[0].shape[0]} rows, {df[0].shape[1]} columns")
    print(f"Columns: {df[0].columns.tolist()}")
    print(f"Data types: {df[0].dtypes.to_dict()}")
    print(df[0].head())
    print(f"Missing values: {df[0].isnull().sum().to_dict()}")
    print(f"Encoding: {df[0].encoding if hasattr(df[0], 'encoding') else 'unknown'}")
    print(f"end=--- {df[1]} ---\n")
```
